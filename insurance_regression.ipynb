# Supervised Learning Task 1: Simple Linear Regression

```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# Load the dataset
df = pd.read_csv("insurance.csv")
df.head()
```

## 1. Exploratory Scatter Plot

We plot age against charges to check whether a linear relationship exists.
Linear regression only makes sense if the data shows a roughly linear trend.

```python
# Plot age vs insurance charges
plt.scatter(df["age"], df["charges"], color="steelblue", alpha=0.5)
plt.xlabel("Age")
plt.ylabel("Charges")
plt.title("Age vs Insurance Charges")
plt.show()
```

## 2. Fit the Regression Model

We extract age as our input (X) and charges as our target (y). Sklearn
requires X to be a 2D array, so we reshape it first. The model finds the
best-fit line by minimising the gap between predicted and actual charges.

```python
# Reshape age column into 2D array for sklearn
X = df["age"].values.reshape(-1, 1)
y = df["charges"].values

# Fit the linear regression model
model = LinearRegression()
model.fit(X, y)

# Generate predictions
y_pred = model.predict(X)
```

## 3. Plot the Regression Line

We overlay the model predictions on the scatter plot as a red line. This
shows how well age alone predicts insurance charges.

```python
# Plot scatter with best-fit line
plt.scatter(df["age"], df["charges"], color="steelblue", alpha=0.5,
            label="Actual")
plt.plot(X, y_pred, color="red", linewidth=2, label="Best-fit line")
plt.xlabel("Age")
plt.ylabel("Charges")
plt.title("Age vs Insurance Charges with Regression Line")
plt.legend()
plt.show()
```
